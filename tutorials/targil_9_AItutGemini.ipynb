```python
import google.generativeai as genai

genai.configure(api_key="")
model = genai.GenerativeModel("gemini-2.5-flash")
```

```python
import re

# Initialize the Gemini model for the chatbot
# This assumes genai.configure() has been run in a previous cell.
chatbot_model = genai.GenerativeModel('gemini-2.5-flash')

# Define patterns and responses for the chatbot, focused on plant diseases
# Expand this dictionary with more plant disease information!
patterns = {
    r"hi|hello|hey": "Hello! I'm here to help you identify and learn about plant diseases. What plant are you observing, or what symptoms are you seeing?",
    r"how are you": "I'm an AI, so I don't have feelings, but I'm ready to assist you with plant disease queries!",
    r"what is your name": "I am a plant disease identification assistant, powered by Google's Gemini model.",
    r"bye|goodbye": "Goodbye! I hope your plants stay healthy!",
    r"tomato (.*) spots": "Reddish-brown spots on tomato leaves might indicate Early Blight or Septoria Leaf Spot. Do you see any yellow halos around the spots or small black dots within them?",
    r"rose (.*) powdery": "Powdery mildew on roses appears as white, powdery patches on leaves and stems. Ensure good air circulation and consider a fungicide treatment.",
    r"yellowing leaves (.*) plant": "Yellowing leaves can be a sign of many issues: overwatering, nutrient deficiency (like nitrogen), or even a disease. Can you tell me more about the plant and how you're caring for it?",
    r"wilting (.*) cucumber": "Wilting in cucumbers could be bacterial wilt, Fusarium wilt, or even just heat stress. Check the stem for any oozing or discoloration."
}

def get_chatbot_response(user_input):
    user_input = user_input.lower()

    # Phase 1 & 2: Pattern matching for quick, predefined responses
    for pattern, response in patterns.items():
        if re.search(pattern, user_input):
            return response

    # Phase 3 (Prompt): Use Gemini for more general or complex queries
    # This prompt is tailored to your plant disease identification project.
    project_context = (
        "You are an expert agricultural bot specializing in plant disease identification and advice. "
        "Provide concise and helpful information based on the user's symptoms or plant type. "
        "If you don't know the answer, politely state that you are still learning or suggest looking for more specific resources. "
        "Focus only on plant-related health and diseases."
    )

    full_prompt = f"{project_context}\nUser: {user_input}\nAssistant:"

    try:
        response = chatbot_model.generate_content(full_prompt)
        return response.text
    except Exception as e:
        return f"I apologize, but I encountered an error while processing your request: {e}. Please ensure your API key is correctly configured and try again later."

print("Plant Disease Chatbot initialized. Type 'quit' to exit.\n")
```

```python
# Demonstrate the chatbot interaction
print("Start chatting with the Plant Disease Bot! Type 'quit' to end.")
while True:
    user_message = input("You: ")
    if user_message.lower() == 'quit':
        break
    bot_response = get_chatbot_response(user_message)
    print(f"Bot: {bot_response}")
print("Chatbot session ended. Happy gardening!")
```
